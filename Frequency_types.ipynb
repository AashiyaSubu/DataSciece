```python
def freqTable(columnName,dataset):
# Get value counts cleanly
    counts = dataset[columnName].value_counts()
    
    # Build dataframe directly from the counts
    freqTable = pd.DataFrame({
        "unique_values": counts.index,
        "Frequency": counts.values,
        "Relative_Frequency": counts.values / len(dataset)
    })
    
    # Add cumulative sum
    freqTable["CumSum"] = freqTable["Relative_Frequency"].cumsum()
    
    return freqTable
```

```python
def quanQual(dataset):
    quan=[]
    qual=[]
    for columnName in dataset.columns:
        #print(columnName)
        if(dataset[columnName].dtype=='O'):
            #print("qual")
            qual.append(columnName)
        else:
            #print("quan")
            quan.append(columnName)
    return quan,qual
```

```python
quan,qaul=quanQual(dataset)
```

```python
dataset[quan]
```

```python
import numpy as np
from sklearn.impute import SimpleImputer
imp = SimpleImputer(missing_values=np.nan, strategy='mean')
imp.fit(dataset[quan])
df=imp.transform(dataset[quan])
df = pd.DataFrame(df,columns = quan)
```

```python
cate=dataset[qaul]
```

```python
two=[df,cate]
```
